```python
def each_class_counts(folder_path):
    classes ={}
    for subdir in os.listdir(folder_path):
        subdir_path = os.path.join(folder_path, subdir)
        file_counts = len(os.listdir(subdir-path))
        classes[subdir_path] = file_counts

    return classes
```
